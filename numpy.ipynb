Create vectors and matrices, compare stepped and evenly spaced sequences, and generate a repeatable sample of random values.

```python
import numpy as np
```

```python
arr = np.array([1, 2, 3])
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
print(arr_2d)
```

```python
print(np.arange(0, 10, 2))  # Step size of 2
print(np.linspace(0, 1, 5))  # Five evenly spaced values, including both endpoints
rng = np.random.default_rng(42)
print(rng.random((3, 2)))
```
